# Kernel PCA

```python
from sklearn.datasets import make_moons
import numpy as np
import matplotlib.pyplot as plt
```

```python
X, y = make_moons(
    n_samples=500,
    noise=0.1,
    random_state=42
    )

plt.scatter(X[:, 0], X[:, 1], c=y)
plt.title("Orginal nonlinear data: Moons")
plt.show()
```

## PCA

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=1)
X_pca = pca.fit_transform(X)
```

```python
plt.figure(figsize=(8, 2))

plt.scatter(
    X_pca[:, 0],
    np.zeros_like(X_pca[:, 0]),
    c=y,
    )

plt.xlabel("Principal Component 1")
plt.yticks([])
plt.title("Linear PCA: 2D → 1D")
plt.show()
```

## Kernel PCA

- Small gamma → smoother/global influence
- Large gamma → more local/flexible influence

```python
from sklearn.decomposition import KernelPCA

kpca = KernelPCA(
    n_components=1,
    kernel="linear",
    gamma=15
    )
X_kpca = kpca.fit_transform(X)
```

```python
plt.figure(figsize=(8, 2))
plt.scatter(
    X_kpca[:, 0],
    np.zeros(len(X_kpca)),
    c=y
)

plt.yticks([])
plt.xlabel("Kernel Principal Component 1")
plt.title("Linear Kernel PCA: 2D → 1D")
plt.show()
```

```python
kpca = KernelPCA(
    n_components=1,
    kernel="rbf",
    gamma=15
    )
X_kpca = kpca.fit_transform(X)

plt.figure(figsize=(8, 2))
plt.scatter(
    X_kpca[:, 0],
    np.zeros(len(X_kpca)),
    c=y
)

plt.yticks([])
plt.xlabel("Kernel Principal Component 1")
plt.title("RBF Kernel PCA: 2D → 1D")
plt.show()
```

```python
kpca = KernelPCA(
    n_components=1,
    kernel="sigmoid",
    gamma=15
    )
X_kpca = kpca.fit_transform(X)

plt.figure(figsize=(8, 2))
plt.scatter(
    X_kpca[:, 0],
    np.zeros(len(X_kpca)),
    c=y
)

plt.yticks([])
plt.xlabel("Kernel Principal Component 1")
plt.title("Sigmoid Kernel PCA: 2D → 1D")
plt.show()
```

```python
kpca = KernelPCA(
    n_components=1,
    kernel="poly",
    gamma=15
    )
X_kpca = kpca.fit_transform(X)

plt.figure(figsize=(8, 2))
plt.scatter(
    X_kpca[:, 0],
    np.zeros(len(X_kpca)),
    c=y
)

plt.yticks([])
plt.xlabel("Kernel Principal Component 1")
plt.title("Polynomial Kernel PCA: 2D → 1D")
plt.show()
```
